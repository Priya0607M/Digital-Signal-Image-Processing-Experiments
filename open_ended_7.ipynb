```python
#import libraries
#cv2 lib reads in this format : blue-green-red plane (bgr)

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#read the input image1

image1 = cv2.imread("images/Picture1.png")
if image1 is None:
    print("Error:Image not found")
    exit()
```

```python
#display the input image1

plt.figure(figsize=(4,4))
plt.imshow(image1,cmap='gray')
plt.title("Input image")
plt.axis("off")
plt.show()
```

```python
#CONSTRAST STRECHING

image1_float = image1.astype(np.float32)

low = np.percentile(image1_float,2)
high = np.percentile(image1_float,98)
stretched = ((image1_float-low) * 255 / (high - low))
stretched = np.clip(stretched,0,255).astype(np.uint8)

#display the output image

plt.figure(figsize=(4,4))
plt.imshow(stretched,cmap='gray',vmin=0,vmax=255)
plt.title("Stretched image")
plt.axis("off")
plt.show()
```

```python
#read the input image2

image2 = cv2.imread("images/Picture2.png")
if image2 is None:
    print("Error:Image not found")
    exit()
```

```python
#display the input image2

plt.figure(figsize=(4,4))
plt.imshow(image2,cmap='gray')
plt.title("Input image")
plt.axis("off")
plt.show()
```

```python
#power law

gamma = 1.5
gamma_image = np.array(255*(image2/255) ** gamma,dtype=np.uint8)

plt.figure(figsize=(4,4))
plt.imshow(gamma_image,cmap="gray")
plt.axis("off")
plt.show()
```
