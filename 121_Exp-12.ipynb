```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "images/BROKEN_IMAGE.png"
image = cv2.imread(image_path,0)
```

```python

_,binary_image = cv2.threshold(image,123,255,cv2.THRESH_BINARY)
```

```python
plt.imshow(image,cmap="gray")
```

```python
plt.imshow(binary_image)
```

```python
kernel = np.array([[-1, 1, -1],
                   [1, 1, 1],
                   [-1, 1, -1]],
                dtype=np.int8)
```

```python
hitmiss_image = cv2.morphologyEx(image, cv2.MORPH_HITMISS, kernel)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(image,cmap="gray")

plt.subplot(1,2,2)
plt.title("HITMISS Image")
plt.imshow(hitmiss_image,cmap="gray")

plt.show()
```

```python
hitmiss_image = cv2.morphologyEx(binary_image, cv2.MORPH_HITMISS, kernel)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(binary_image,cmap="gray")

plt.subplot(1,2,2)
plt.title("HITMISS Image")
plt.imshow(hitmiss_image,cmap="gray")

plt.show()
```
