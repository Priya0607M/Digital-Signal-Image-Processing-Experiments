```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "images/BROKEN_IMAGE.png"
image = cv2.imread(image_path,0)
```

```python

_,binary_image = cv2.threshold(image,123,255,cv2.THRESH_BINARY)
```

```python
plt.imshow(image,cmap="gray")
```

```python
plt.imshow(binary_image)
```

```python
kernel = cv2.getStructuringElement(cv2.MORPH_RECT,(3,3))
print(kernel)
```

```python

dilated_image = cv2.dilate(image,kernel,iterations=3)
eroded_image = cv2.erode(image,kernel,iterations=3)
opened_image = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)
closed_image = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,5,1)
plt.title("Original Image")
plt.imshow(image,cmap="gray")

plt.subplot(1,5,2)
plt.title("Dilated Image")
plt.imshow(dilated_image,cmap="gray")

plt.subplot(1,5,3)
plt.title("Eroded Image")
plt.imshow(eroded_image,cmap="gray")

plt.subplot(1,5,4)
plt.title("Opened Image")
plt.imshow(opened_image,cmap="gray")

plt.subplot(1,5,5)
plt.title("Closed Image")
plt.imshow(closed_image,cmap="gray")

plt.show()
```

```python

dilated_image = cv2.dilate(binary_image,kernel,iterations=3)
eroded_image = cv2.erode(binary_image,kernel,iterations=3)
opened_image = cv2.morphologyEx(binary_image, cv2.MORPH_OPEN, kernel)
closed_image = cv2.morphologyEx(binary_image, cv2.MORPH_CLOSE, kernel)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,5,1)
plt.title("Original Image")
plt.imshow(binary_image,cmap="gray")

plt.subplot(1,5,2)
plt.title("Dilated Image")
plt.imshow(dilated_image,cmap="gray")

plt.subplot(1,5,3)
plt.title("Eroded Image")
plt.imshow(eroded_image,cmap="gray")

plt.subplot(1,5,4)
plt.title("Opened Image")
plt.imshow(opened_image,cmap="gray")

plt.subplot(1,5,5)
plt.title("Closed Image")
plt.imshow(closed_image,cmap="gray")


plt.show()
```
