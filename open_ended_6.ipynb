```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read Image 1
image1 = cv2.imread("images/Image_1.png", 0)

# Read Image 2
image2 = cv2.imread("images/Image_2.png", 0)

# Read Image 3
image3 = cv2.imread("images/Image_3.png", 0)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(image1, cmap="gray")
plt.title("Image 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(image2, cmap="gray")
plt.title("Image 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(image3, cmap="gray")
plt.title("Image 3")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Kernel for Image 1
kernel1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (35, 35))

# Binary conversion
_, binary1 = cv2.threshold(image1, 127, 255,cv2.THRESH_BINARY)

# Invert image
inverted1 = cv2.bitwise_not(binary1)

# Apply closing
closed1 = cv2.morphologyEx(inverted1,cv2.MORPH_CLOSE,kernel1)

# Invert back
result1 = cv2.bitwise_not(closed1)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title("Original Image 1")
plt.imshow(image1, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Inverted Image")
plt.imshow(inverted1, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Final Image")
plt.imshow(result1, cmap="gray")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Kernel for Image 2
kernel2_open = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9, 3))

kernel2_close = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))

# Apply opening to remove small bright noise
opened2 = cv2.morphologyEx(image2,cv2.MORPH_OPEN,kernel2_open)

# Apply closing to fill narrow dark cracks
result2 = cv2.morphologyEx(opened2,cv2.MORPH_CLOSE,kernel2_close)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title("Original Image 2")
plt.imshow(image2, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Opened Image")
plt.imshow(opened2, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Final Image")
plt.imshow(result2, cmap="gray")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Kernel for Image 3
kernel3 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))

# Binary conversion
_, binary3 = cv2.threshold(image3, 127, 255,cv2.THRESH_BINARY)

# Remove small bright noise
opened3 = cv2.morphologyEx(binary3,cv2.MORPH_OPEN,kernel3)

# Fill narrow dark cracks
result3 = cv2.morphologyEx(opened3,cv2.MORPH_CLOSE,kernel3)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title("Original Image 3")
plt.imshow(image3, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Opened Image")
plt.imshow(opened3, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Final Image")
plt.imshow(result3, cmap="gray")
plt.axis("off")

plt.tight_layout()
plt.show()
```
