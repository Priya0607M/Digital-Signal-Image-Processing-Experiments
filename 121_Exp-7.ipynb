```python
#import libraries
#cv2 lib reads in this format : blue-green-red plane (bgr)

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#read the input image

image = cv2.imread("images/image_1.jpg")
if image is None:
    print("Error:Image not found")
    exit()
```

```python
#this will give the original image --> instead of reading in bgr plane it will read in rgb

image1 = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
#display the input image

plt.figure(figsize=(4,4))
plt.imshow(image1)
plt.title("Input image")
plt.axis("off")
plt.show()
```

```python
gray_image = cv2.imread("images/image_1.jpg",cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(4,4))
plt.imshow(gray_image,cmap="gray") #default-verdis : yellowish shade
plt.title("Gray image")
plt.axis("off")
plt.show()
```

```python
negat_image = 255 - gray_image

plt.figure(figsize=(4,4))
plt.imshow(negat_image,cmap="gray") #default-verdis : yellowish shade
plt.title("Negat image")
plt.axis("off")
plt.show()
```

```python
# c = 255 / np.log(1 + np.max(gray_image))
# print(c)
```

```python
#apply log transformation
c = 40
log_image = c * np.log(1+gray_image.astype(np.float32))
log_image = np.array(log_image,dtype=np.uint8)

plt.figure(figsize=(4,4))
plt.imshow(log_image,cmap="gray") #default-verdis : yellowish shade
plt.title("Log image")
plt.axis("off")
plt.show()
```

```python
#inverse log tranformation

```

```python
#power law

gamma = 0.5 #gamma>1--dark ; gamma<1--brighter ; gamma =1--same
gamma_image = np.array(255*(gray_image/255) ** gamma,dtype=np.uint8)

plt.figure(figsize=(4,4))
plt.imshow(gamma_image,cmap="gray") #default-verdis : yellowish shade
plt.title("Gamma image")
plt.axis("off")
plt.show()
```

```python
threshold_value = 80  #remove further noise / unwanted information
thresh_value, thresh_image = cv2.threshold(gray_image,threshold_value,255,cv2.THRESH_BINARY)
#image and value will be provided if not error
plt.figure(figsize=(4,4))
plt.imshow(thresh_image,cmap="gray") #default-verdis : yellowish shade
plt.title("Threshold image")
plt.axis("off")
plt.show()
```

```python
#ADAPTIVE THRESHOLDING (GUASSIAN)

# threshold_value = 80  #remove further noise / unwanted information
thresh_image = cv2.adaptiveThreshold(gray_image,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
#image and value will be provided if not error
plt.figure(figsize=(4,4))
plt.imshow(thresh_image,cmap="gray") #default-verdis : yellowish shade
plt.title("Adaptive (GUASSIAN) Threshold image")
plt.axis("off")
plt.show()
```

```python
#ADAPTIVE THRESHOLDING (MEAN)

# threshold_value = 80  #remove further noise / unwanted information
thresh_image = cv2.adaptiveThreshold(gray_image,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,11,2)
#image and value will be provided if not error
plt.figure(figsize=(4,4))
plt.imshow(thresh_image,cmap="gray") #default-verdis : yellowish shade
plt.title("Adaptive (MEAN) Threshold image")
plt.axis("off")
plt.show()
```

```python
#OTSU THRESHOLDING 

# threshold_value = 80  #remove further noise / unwanted information
_,thresh_image = cv2.threshold(gray_image,0,255,cv2.THRESH_BINARY + cv2.THRESH_OTSU)
#image and value will be provided if not error
plt.figure(figsize=(4,4))
plt.imshow(thresh_image,cmap="gray") #default-verdis : yellowish shade
plt.title("OTSU Threshold image")
plt.axis("off")
plt.show()
```
