```python
#import libraries
#cv2 lib reads in this format : blue-green-red plane (bgr)

import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.exposure import match_histograms
```

```python
def generate_histogram(image):

    histogram = cv2.calcHist(
        [image], 
        [0], 
        None, 
        [256], 
        [0, 256])
    
    return histogram
```

```python
def perform_histogram_equalization(image):
  
  gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
  equalized_image = cv2.equalizeHist(gray_image) # Apply histogram equalization
  
  return equalized_image
```

```python
def perform_histogram_matching(input_image, reference_image):

  gray_input = cv2.cvtColor(input_image, cv2.COLOR_BGR2GRAY)

  gray_reference = cv2.cvtColor(reference_image, cv2.COLOR_BGR2GRAY)

  input_hist = generate_histogram(gray_input) # Calculate histograms
  reference_hist = generate_histogram(gray_reference)

  matched_image = match_histograms(gray_input, gray_reference)# Perform histogram matching
  matched_image = np.uint8(matched_image)
  
  return matched_image
```

```python
# Load the input image 1
input_path = 'images/Picture1.png'
input_image = cv2.imread(input_path)

# Generate and display the histogram of the input image
gray_input = cv2.cvtColor(input_image,cv2.COLOR_BGR2GRAY)
input_hist = generate_histogram(gray_input)
```

```python

# Perform histogram equalization
equalized_image = perform_histogram_equalization(input_image)
# Generate and display the histogram of the equalized image
equalized_hist = generate_histogram(equalized_image)
```

```python
# Load the reference image for histogram matching
reference_path = 'images/images.png'
reference_image = cv2.imread(reference_path)
# Perform histogram matching
matched_image = perform_histogram_matching(input_image, reference_image)
# Generate and display the histogram of the matched image
matched_hist = generate_histogram(matched_image)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.plot(input_hist)
plt.title("Input Image Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.subplot(1,3,2)
plt.plot(equalized_hist)
plt.title("Equalized Image Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.subplot(1,3,3)
plt.plot(matched_hist)
plt.title("Matched Image Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(gray_input,cmap='gray')
plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(equalized_image,cmap='gray')
plt.title("Histogram Equalization")

plt.subplot(1,3,3)
plt.imshow(matched_image,cmap='gray')
plt.title("Histogram Matching")

plt.tight_layout()
plt.show()
```

```python

# Save the equalized and matched images 
equalized_path = 'equalized_image1.jpg'
matched_path = 'matched_image1.jpg'

cv2.imwrite(equalized_path, equalized_image)
cv2.imwrite(matched_path, matched_image)
```
