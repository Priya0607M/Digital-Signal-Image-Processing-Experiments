```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def mean_filter(image):
    mean_3x3 = cv2.blur(image, (3,3))
    mean_5x5 = cv2.blur(image, (5,5))

    return mean_3x3, mean_5x5

def weighted_mean_filter(image):

    weighted_kernel = np.array([
        [1,1,2],
        [1,3,1],
        [1,1,7]
    ], dtype=np.float32)

    weighted_kernel = weighted_kernel / 18

    weighted_mean = cv2.filter2D(
        image,
        -1,
        weighted_kernel
    )

    return weighted_mean


def gaussian_filter(image):

    gaussian_kernel = np.array([
        [1,2,1],
        [2,4,2],
        [1,2,1]
    ], dtype=np.float32)

    gaussian_kernel = gaussian_kernel / 16

    gaussian_mean = cv2.filter2D(
        image,
        -1,
        gaussian_kernel
    )

    return gaussian_mean


def laplacian_sharpening(image):

    laplacian = cv2.Laplacian(
        image,
        cv2.CV_32F
    )

    laplacian_abs = cv2.convertScaleAbs(
        laplacian
    )

    sharpened = cv2.convertScaleAbs(
        image.astype(np.float32) - laplacian
    )

    return laplacian_abs, sharpened
```

```python
image1 = cv2.imread("images/IMAGE1.png")

image1_rgb = cv2.cvtColor(
    image1,
    cv2.COLOR_BGR2RGB
)

mean1_3x3, mean1_5x5 = mean_filter(image1)

weighted1 = weighted_mean_filter(image1)

gaussian1 = gaussian_filter(image1)

laplacian1, sharpened1 = laplacian_sharpening(image1)

plt.figure(figsize=(15,8))

plt.subplot(2,3,1)
plt.imshow(image1_rgb)
plt.title("Original Image 1")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(
    cv2.cvtColor(mean1_3x3, cv2.COLOR_BGR2RGB)
)
plt.title("Mean Filter 3x3")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(
    cv2.cvtColor(mean1_5x5, cv2.COLOR_BGR2RGB)
)
plt.title("Mean Filter 5x5")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(
    cv2.cvtColor(weighted1, cv2.COLOR_BGR2RGB)
)
plt.title("Weighted Mean")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(
    cv2.cvtColor(gaussian1, cv2.COLOR_BGR2RGB)
)
plt.title("Gaussian Filter")
plt.axis("off")

plt.tight_layout()
plt.show()

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(image1_rgb)
plt.title("Original Image 1")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(laplacian1, cmap="gray")
plt.title("Laplacian Edge Detection")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(
    cv2.cvtColor(sharpened1, cv2.COLOR_BGR2RGB)
)
plt.title("Laplacian Sharpened")
plt.axis("off")

plt.tight_layout()
plt.show()
```
